**Taller Big Data**

```python
#Instalación de bibliotecas  - Se importan las bibliotecas para utilizar sus funciones
import pandas as pd
import numpy as np
```

```python
#Números aleatorios paa generación de tabla
np.random.seed(42)
n = 200
```

```python
#Creación de tabala de ventas
Ventas = pd.DataFrame({
    "fecha": pd.date_range("2026-01-01", periods=n, freq="D"),
    "ciudad": np.random.choice(["Bogotá", "Medellín", "Cali", "Barranquilla"], n),
    "canal": np.random.choice(["Tienda", "Web", "App"], n),
    "unidades": np.random.randint(1, 20, n),
    "precio": np.random.choice([25000, 48000, 99000], n),
})
```

```python
#Adición de columna de total en tabla Ventas
Ventas["Total"] = Ventas["unidades"] * Ventas["precio"]
```

```python
#Imprimir resultados
print(Ventas.shape)
print(Ventas.head(3))
```

```python
#Mostrar columnas y tipos de datos
Ventas.info()
```

**Analítica Descriptiva: ¿Qué pasó?**

```python
#Generar una tabla dinámica con los datos
resumen = (Ventas.groupby("ciudad")["Total"]
                 .agg(["count", "sum", "mean"])
                 .sort_values("sum", ascending=False))

```

```python
#Resumen de datos - Mostrar en millones el total y postrar el promedio
resumen["sum"] = (resumen["sum"] / 1e6).round(1)
resumen["mean"] = resumen["mean"].round(0)
resumen.columns = ["Ventas", "Total_millones", "Ticket_prom"]
```

```python
# Visualizar
print(resumen)
```
