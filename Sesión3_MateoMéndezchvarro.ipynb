```python
# Problema 3: ¿te alcanza para el café?

presupuesto = 500000   # víveres al  mes
precio_cafe = 5000
cafes_semana = 0    # No tomo café

# 1. ¿Cuánto gastas al mes? (4 semanas)
gasto_mes = precio_cafe * cafes_semana * 4

# 2. ¿Cuánto te sobra?
sobra = presupuesto - gasto_mes

# 3. ¿Te alcanza? (True o False)
alcanza = presupuesto >= gasto_mes

print(gasto_mes, sobra, alcanza)
```

```python
# Problema 7: ¿te toca envío gratis?

total = 850
es_miembro = True
compras = 7

# 1. Envío: más de 1000 → gratis; miembro → gratis; si no → $99
if total > 1000:
    print("Envío gratis por monto")
elif es_miembro:
    print("Envío gratis por ser miembro")
else:
    print("Envío: $99")

# 2. Nivel: 10 o más → Oro; de 5 a 9 → Plata; menos → Bronce
if compras >= 10:
    print("Nivel Oro")
elif compras >= 5:
    print("Nivel Plata")
else:
    print("Nivel Bronce")
```
