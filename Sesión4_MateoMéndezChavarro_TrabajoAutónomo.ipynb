```python
# Ejercicio 3: primer análisis de ventas

#Instalación de bibliotecas  - Se importan las bibliotecas para utilizar sus funciones
import pandas as pd
import numpy as np
#Números aleatorios paa generación de tabla
np.random.seed(42)
n = 200
#Creación de tabala de ventas
Ventas = pd.DataFrame({
    "fecha": pd.date_range("2026-01-01", periods=n, freq="D"),
    "ciudad": np.random.choice(["Bogotá", "Medellín", "Cali", "Barranquilla"], n),
    "canal": np.random.choice(["Tienda", "Web", "App"], n),
    "unidades": np.random.randint(1, 20, n),
    "precio": np.random.choice([25000, 48000, 99000], n),
})
#Adición de columna de total en tabla Ventas
Ventas["Total"] = Ventas["unidades"] * Ventas["precio"]
#Imprimir resultados
print(Ventas.shape)
print(Ventas.head(3))
#Mostrar columnas y tipos de datos
Ventas.info()
#Generar una tabla dinámica con los datos
resumen = (Ventas.groupby("ciudad")["Total"]
                 .agg(["count", "sum", "mean"])
                 .sort_values("sum", ascending=False))
#Resumen de datos - Mostrar en millones el total y postrar el promedio
resumen["sum"] = (resumen["sum"] / 1e6).round(1)
resumen["mean"] = resumen["mean"].round(0)
resumen.columns = ["Ventas", "Total_millones", "Ticket_prom"]

# 1. Total vendido por canal
print(Ventas.groupby("canal")["precio"].sum())

# 2. Ticket promedio por ciudad
print(Ventas.groupby("ciudad")["precio"].mean())

# 3. Día de la semana en que más se vende
Ventas["dia"] = Ventas["fecha"].dt.day_name()
por_dia = Ventas.groupby("dia")["precio"].sum().sort_values(ascending=False)
print(por_dia)
print("Día con más ventas:", por_dia.idxmax())
```

La tienda física concentra la mayor parte de las ventas y Barranquilla tiene el ticket más alto, así que conviene reforzar inventario en esa tienda

```python
# Ejercicio 1: clasificador de notas

def clasificar_nota(nota):
    if nota < 0 or nota > 5:
        return "Nota inválida"
    elif nota < 3.0:
        return "Reprueba"
    elif nota < 4.5:
        return "Aprueba"
    else:
        return "Excelente"

notas = [2.5, 3.8, 4.7, 5.2, 3.0]
aprobados = 0

for nota in notas:
    estado = clasificar_nota(nota)
    print(nota, "→", estado)
    if estado == "Aprueba" or estado == "Excelente":
        aprobados = aprobados + 1

print("Estudiantes que aprobaron:", aprobados)
```

```python
# Ejercicio 2: carrito de compras

carrito = {
    "camiseta": (45000, 2),
    "jean": (120000, 1),
    "medias": (12000, 3),
}

# Subtotal
subtotal = 0
for producto, (precio, cantidad) in carrito.items():
    subtotal = subtotal + precio * cantidad

# Descuento del 10% si supera 200.000
descuento = 0
if subtotal > 200000:
    descuento = subtotal * 0.10

# IVA del 19% sobre el valor ya descontado
base = subtotal - descuento
iva = base * 0.19
total = base + iva

def pesos(valor):
    return "$" + f"{valor:,.0f}".replace(",", ".")

print("Subtotal:", pesos(subtotal))
print("Descuento:", pesos(descuento))
print("IVA (19%):", pesos(iva))
print("Total a pagar:", pesos(total))
```

```python
# Ejercicio 3: nómina con pandas
import pandas as pd

# Datos de ejemplo: si ya tienes el DataFrame "empleados" de la actividad, usa ese
empleados = pd.DataFrame({
    "nombre": ["Ana", "Luis", "Marta", "Carlos", "Sofía"],
    "area": ["Ventas", "Finanzas", "Ventas", "Operaciones", "Finanzas"],
    "salario": [3200000, 4500000, 2800000, 3600000, 5200000],
})

# 1. Agregar 3 empleados nuevos
nuevos = pd.DataFrame({
    "nombre": ["Diego", "Laura", "Andrés"],
    "area": ["Ventas", "Operaciones", "Finanzas"],
    "salario": [3000000, 3900000, 4800000],
})
empleados = pd.concat([empleados, nuevos], ignore_index=True)

# 2. Bono: 10% si es Ventas, 5% si no
empleados["bono"] = empleados.apply(
    lambda fila: fila["salario"] * 0.10 if fila["area"] == "Ventas" else fila["salario"] * 0.05,
    axis=1,
)

# 3. Costo total de nómina mensual por área (salario + bono)
empleados["costo"] = empleados["salario"] + empleados["bono"]
print(empleados.groupby("area")["costo"].sum())

# 4. Empleado con mayor salario
mejor_pagado = empleados.loc[empleados["salario"].idxmax()]
print("Mayor salario:", mejor_pagado["nombre"], "-", mejor_pagado["salario"])
```
